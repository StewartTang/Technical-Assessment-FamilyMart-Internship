Build a solution to transcribe audio to text, especially in Malaysian
spoken languages.

Installing necessary packages needed for this tasks.

```python
pip install -U openai-whisper
```

Import the whisper model.

```python
import whisper
```

Used large whisper model to transcribe audio into text.

```python
model = whisper.load_model("large")
```

Use the model to transcribe the Sample_Audio.mp3 file.


```python
audio = model.transcribe("Sample_Audio.mp3")
```

Print the audio text.

```python
print(audio["text"])
```

Output the audio text into transcription_output.txt file.

```python
with open("transcription_output.txt", "w", encoding="utf-8") as f:
  f.write(audio["text"])
```
